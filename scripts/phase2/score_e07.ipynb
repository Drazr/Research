# score_e07
Phase 2 Stage 4 E07 CPU frozen-head scorer and dual-cohort alternate-caption comparator.

- Configure only cell 3 (`MODEL = 'qwen'` or `'gemma'`, `PLAN_REF`, `E07_AUDIT_REF`, `E07_RELEASE_REF`, `E29_ANALYSIS_REF`, optional resume refs).
- Uses the pinned Python 3.11/3.12 CPU runtime (`requirements-stage2-cpu.txt`) with single-threaded BLAS/OpenMP.
- Authenticates `phase2_plan`, `phase2_e07_audit`, `phase2_e07_release` (all chunks + `.npy` hashes + repeat checks), and Stage 2 E29 `draw0_reference`; runs `en->en` draw-0 reproduction self-checks (`atol=0, rtol=0`); scores ONLY verified unique changed views with the 8 frozen Stage 3 midpoint `image_caption` heads/scalers (handling zero-changed languages like `bn` gracefully); reuses Stage 4 draw-0 logits bit-for-bit on unchanged/byte-identical rows; and publishes both `full_screen_360` (12-route + 9-route changed-targets) and `restricted_changed_cohort` (exact matched draw-0 companion on sliced `(2000, 360)` bootstrap counts).
- Phase 1 directories remain strictly read-only, heads are never refit, and `CONFIRM` stays closed.


```python
import os, sys, subprocess
from pathlib import Path

if sys.version_info[:2] not in ((3, 11), (3, 12)):
    raise RuntimeError('Use Python 3.11/3.12 with the pinned CPU requirements.')
for name in ('OPENBLAS_NUM_THREADS', 'OMP_NUM_THREADS', 'MKL_NUM_THREADS', 'NUMEXPR_NUM_THREADS'):
    os.environ[name] = '1'

from google.colab import drive
drive.mount('/content/drive')

REPO = Path('/content/Research-phase2-score-e07')
CODE_REF = 'phase2'  # Set to the approved Phase 2 branch or commit SHA to pin explicitly.
if not REPO.exists():
    subprocess.run(['git', 'clone', 'https://github.com/Drazr/Research.git', str(REPO)], check=True)
if subprocess.check_output(['git', 'status', '--porcelain'], cwd=REPO, text=True).strip():
    raise RuntimeError('Use a clean checkout; preserve local work before continuing.')
subprocess.run(['git', 'fetch', 'origin', CODE_REF], cwd=REPO, check=True)
CODE_COMMIT = subprocess.check_output(['git', 'rev-parse', 'FETCH_HEAD'], cwd=REPO, text=True).strip()
subprocess.run(['git', 'checkout', '--detach', CODE_COMMIT], cwd=REPO, check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', str(REPO / 'requirements-stage2-cpu.txt')], check=True)
print('Fixed code commit:', CODE_COMMIT)
print('Restart if installation requests it, then rerun setup before continuing.')
```

```python
# Visible Stage 4 E07 CPU scoring settings (edit only this cell).
PHASE1_ROOT = Path('/content/drive/MyDrive/phase1')
MODEL = 'qwen'               # 'qwen' or 'gemma' (run one CPU runtime per model).
PLAN_REF = None              # Required: {'path': 'phase2/plans/.../_COMPLETE.json', 'sha256': '...'} or Path.
E07_AUDIT_REF = None         # Required: {'path': 'phase2/audits/e07/.../_COMPLETE.json', 'sha256': '...'} or Path.
E07_RELEASE_REF = None       # Required: model's {'path': 'phase2/features/e07/<model>/releases/.../_COMPLETE.json', 'sha256': '...'} or Path.
E29_ANALYSIS_REF = None      # Required: model's {'path': 'phase2/results/e29/<model>/.../_COMPLETE.json', 'sha256': '...'} or Path.
RESUME_PREDICTION_REFS = None  # Optional dict {'last_caption:en': {'path': ..., 'sha256': ...}, ...} or JSON file path.
RESUME_ANALYSIS_REF = None     # Optional explicit {'path': ..., 'sha256': ...} or Path to phase2_analysis_block _COMPLETE.json.
OUTPUT_REF = None              # Optional mounted JSON path (e.g. PHASE1_ROOT / f'phase2/results/e07_{MODEL}_ref.json').
```

```python
import json, signal

if subprocess.check_output(['git', 'rev-parse', 'HEAD'], cwd=REPO, text=True).strip() != CODE_COMMIT:
    raise RuntimeError('Checkout changed after setup.')
if subprocess.check_output(['git', 'status', '--porcelain'], cwd=REPO, text=True).strip():
    raise RuntimeError('Use a clean Phase 2 checkout.')
if MODEL not in ('qwen', 'gemma'):
    raise RuntimeError("Set MODEL to 'qwen' or 'gemma'.")
if not PHASE1_ROOT.is_dir():
    raise RuntimeError(f'Missing mounted PHASE1_ROOT directory: {PHASE1_ROOT}')
for name, val in (
    ('PLAN_REF', PLAN_REF),
    ('E07_AUDIT_REF', E07_AUDIT_REF),
    ('E07_RELEASE_REF', E07_RELEASE_REF),
    ('E29_ANALYSIS_REF', E29_ANALYSIS_REF),
):
    if val is None:
        raise RuntimeError(f'Set {name} to the certified Stage 1/2 completion reference.')

def _format_ref(ref, label):
    if isinstance(ref, dict):
        target = PHASE1_ROOT / ref['path']
        if not target.is_file():
            raise RuntimeError(f'Missing {label} manifest: {target}')
        return json.dumps(ref)
    target = Path(ref)
    if not target.is_file():
        raise RuntimeError(f'Missing {label} manifest: {target}')
    return str(target)

command = [
    sys.executable, '-u', '-m', 'src.phase2.score_e07',
    '--phase1-root', str(PHASE1_ROOT),
    '--plan', _format_ref(PLAN_REF, 'phase2_plan'),
    '--e07-audit', _format_ref(E07_AUDIT_REF, 'phase2_e07_audit'),
    '--e07-release', _format_ref(E07_RELEASE_REF, 'phase2_e07_release'),
    '--e29-analysis', _format_ref(E29_ANALYSIS_REF, 'e29_analysis'),
    '--model', MODEL,
    '--storage-scope', 'persistent-drive',
]
if RESUME_PREDICTION_REFS is not None:
    if isinstance(RESUME_PREDICTION_REFS, dict):
        command += ['--resume-prediction-refs', json.dumps(RESUME_PREDICTION_REFS)]
    else:
        command += ['--resume-prediction-refs', str(RESUME_PREDICTION_REFS)]
if RESUME_ANALYSIS_REF is not None:
    command += ['--resume-analysis-ref', _format_ref(RESUME_ANALYSIS_REF, 'resume_analysis')]
if OUTPUT_REF is not None:
    command += ['--output-ref', str(OUTPUT_REF)]

process = subprocess.Popen(
    command, cwd=REPO, env=os.environ.copy(),
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1,
)
blockers = []
events = []
try:
    for line in process.stdout:
        print(line, end='', flush=True)
        try:
            payload = json.loads(line)
            if isinstance(payload, dict):
                events.append(payload)
                if payload.get('status') == 'blocked':
                    blockers.extend(payload.get('blockers', []))
        except json.JSONDecodeError:
            pass
except KeyboardInterrupt:
    process.send_signal(signal.SIGINT)
    for line in process.stdout:
        print(line, end='', flush=True)
    process.wait()
    raise
code = process.wait()
if code != 0:
    detail = '; '.join(blockers) if blockers else f'exit code {code}'
    raise RuntimeError(f'E07 scorer blocked/failed for {MODEL}: {detail}')

completed = [e for e in events if e.get('event') == 'e07_score_complete']
if completed:
    print('\n=== E07 SCORING & ANALYSIS REFERENCES ===')
    print(json.dumps(completed[-1], indent=2))
```
